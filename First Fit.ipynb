```python
def first_fit(block_sizes, process_sizes):
    allocation = [-1] * len(process_sizes)

    for i, process in enumerate(process_sizes):
        for j, block in enumerate(block_sizes):
            if block >= process:
                allocation[i] = j
                block_sizes[j] -= process
                break

    print("Process No.\tProcess Size\tBlock No.")
    for i in range(len(process_sizes)):
        print(f"{i + 1}\t\t{process_sizes[i]}\t\t{allocation[i] + 1 if allocation[i] != -1 else 'Not Allocated'}")

# Example
block_sizes = [100, 500, 200, 300, 600]
process_sizes = [212, 417, 112, 426]
first_fit(block_sizes, process_sizes)
```
